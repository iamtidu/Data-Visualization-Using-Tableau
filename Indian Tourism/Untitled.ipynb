```python
import random
import pandas as pd
from datetime import datetime, timedelta

# List of Indian states and major cities
states_cities = {
    "Uttar Pradesh": ["Agra", "Varanasi", "Lucknow"],
    "Maharashtra": ["Mumbai", "Pune", "Nagpur"],
    "Rajasthan": ["Jaipur", "Udaipur", "Jodhpur"],
    "Kerala": ["Kochi", "Thiruvananthapuram", "Munnar"],
    "Tamil Nadu": ["Chennai", "Madurai", "Coimbatore"],
    "Delhi": ["New Delhi"],
    "Goa": ["Panaji", "Vasco da Gama"],
    "Karnataka": ["Bengaluru", "Mysuru"],
    "West Bengal": ["Kolkata", "Darjeeling"],
    "Gujarat": ["Ahmedabad", "Surat"],
    "Punjab": ["Amritsar", "Ludhiana"],
    "Himachal Pradesh": ["Shimla", "Manali"],
    "Madhya Pradesh": ["Bhopal", "Indore"],
    "Bihar": ["Patna", "Gaya"],
    "Assam": ["Guwahati", "Tezpur"],
    "Odisha": ["Bhubaneswar", "Puri"],
    "Jharkhand": ["Ranchi", "Jamshedpur"],
    "Chhattisgarh": ["Raipur", "Bilaspur"],
    "Haryana": ["Gurgaon", "Faridabad"],
    "Telangana": ["Hyderabad"],
    "Andhra Pradesh": ["Visakhapatnam", "Vijayawada"],
    "Uttarakhand": ["Dehradun", "Haridwar"],
    "Jammu and Kashmir": ["Srinagar", "Jammu"],
    "Sikkim": ["Gangtok"],
    "Tripura": ["Agartala"],
    "Manipur": ["Imphal"],
    "Meghalaya": ["Shillong"],
    "Mizoram": ["Aizawl"],
    "Nagaland": ["Kohima"],
    "Arunachal Pradesh": ["Itanagar"],
    "Puducherry": ["Puducherry"],
    "Chandigarh": ["Chandigarh"],
    "Ladakh": ["Leh"]
}

popular_attractions = [
    "Beach", "Temple", "Fort", "Hill Station", "Wildlife Sanctuary", 
    "Museum", "Cultural Festival", "Backwaters", "Desert Safari", "River Rafting"
]

weather_conditions = ["Sunny", "Rainy", "Cloudy", "Cold", "Humid", "Snowy"]

events_or_festivals = [
    "Festival", "Food Festival", "Music Festival", "Art Fair", "Dance Festival", 
    "Crafts Mela", "Religious Event", "Sports Event"
]

modes_of_transport = ["Flight", "Train", "Bus", "Car", "Bike", "Boat"]

medical_facility_ratings = ["Good", "Average", "Poor"]

# Generate data
start_date = datetime(2019, 1, 1)
end_date = datetime(2023, 12, 31)

data = []

for state, cities in states_cities.items():
    for city in cities:
        current_date = start_date
        while current_date <= end_date:
            domestic_tourists = random.randint(5000, 50000)
            international_tourists = random.randint(500, 5000)
            tourism_revenue = round((domestic_tourists * random.uniform(0.5, 1.5)) + (international_tourists * random.uniform(2.0, 5.0)), 2)
            avg_spending = round(tourism_revenue / (domestic_tourists + international_tourists), 2)
            number_of_hotels = random.randint(50, 500)
            avg_hotel_price = round(random.uniform(1500, 10000), 2)
            attraction = random.choice(popular_attractions)
            event = random.choice(events_or_festivals)
            weather = random.choice(weather_conditions)
            transport_mode = random.choice(modes_of_transport)
            avg_transport_cost = round(random.uniform(500, 8000), 2)
            local_transport_used = random.choice(["Yes", "No"])
            avg_stay_days = round(random.uniform(2, 15), 1)
            hotel_occupancy_rate = round(random.uniform(50, 95), 2)  # %
            tourist_satisfaction = round(random.uniform(3.0, 5.0), 1)  # out of 5
            adventure_activities = random.choice(["Yes", "No"])
            medical_facilities = random.choice(medical_facility_ratings)
            
            data.append({
                "State": state,
                "City": city,
                "Date": current_date.strftime("%Y-%m-%d"),
                "Year": current_date.year,
                "Month": current_date.strftime("%B"),
                "Domestic_Tourists": domestic_tourists,
                "International_Tourists": international_tourists,
                "Tourism_Revenue_Cr": tourism_revenue,
                "Popular_Attraction": attraction,
                "Average_Spending_per_Tourist": avg_spending,
                "Number_of_Hotels": number_of_hotels,
                "Average_Hotel_Price": avg_hotel_price,
                "Events_or_Festivals": event,
                "Weather_Condition": weather,
                "Mode_of_Transport": transport_mode,
                "Average_Transport_Cost": avg_transport_cost,
                "Local_Transport_Used": local_transport_used,
                "Average_Stay_Days": avg_stay_days,
                "Hotel_Occupancy_Rate(%)": hotel_occupancy_rate,
                "Tourist_Satisfaction_Score": tourist_satisfaction,
                "Adventure_Activities": adventure_activities,
                "Medical_Facilities": medical_facilities
            })
            
            # Increment date by 1 month
            if current_date.month == 12:
                current_date = current_date.replace(year=current_date.year + 1, month=1, day=1)
            else:
                current_date = current_date.replace(month=current_date.month + 1, day=1)

# Create DataFrame
df = pd.DataFrame(data)

# Save to CSV
df.to_csv("enhanced_india_tourism_dataset.csv", index=False)

print("Enhanced Dataset generated successfully with", len(df), "rows!")
```
